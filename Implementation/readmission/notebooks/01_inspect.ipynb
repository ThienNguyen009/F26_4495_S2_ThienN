```python
import pandas as pd

df = pd.read_csv("data/diabetic_data.csv", na_values=["?", ""], keep_default_na=False)
print(df.shape)
print(df["readmitted"].value_counts(normalize=True))
print(df.isna().mean().sort_values(ascending=False).head(10))
print(df["patient_nbr"].nunique(), "unique patients")
print(df["patient_nbr"].value_counts().head())     # patients with many encounters
for col in ["A1Cresult", "max_glu_serum", "gender", "race", "age"]:
    print(df[col].value_counts(dropna=False), "\n")
```
